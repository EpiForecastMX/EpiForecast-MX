## 1. Estrategias de ensamble homogéneas y heterogéneas <a id='sec1'></a>

# Homogéneos

- **Prophet** con múltiples configuraciones de hiperparámetros (seasonalidad, Fourier order, holidays).
- **DeepAR** entrenado en diferentes configuraciones de context length y distribución (ej. Student-t vs Gaussian).
- *Ambos se optimizan con time-series cross-validation y grids específicos.*

# Heterogéneos

- **Ensemble Prophet + XGBoost**: Prophet captura tendencia/estacionalidad, XGBoost corrige residuales.
- **Stacking (Prophet + ETS + LightGBM + Ridge)**: combinación de expertos con meta-learner regularizado.


## 2. Optimización de hiperparámetros <a id='sec1'></a>


- Grids por condición

Se definieron grids específicos por patología, basados en resultados previos. Cada grid busca balancear regularización, sensibilidad a cambios estructurales y estacionalidad.

# Prophet

En esta configuración de Prophet, ajustamos el modelo para que se enfoque exclusivamente en detectar patrones anuales a la medida, ignorando variaciones diarias o semanales que no aportan valor. También limitamos su sensibilidad a cambios bruscos para evitar que se confunda con variaciones al azar en estados pequeños, y le enseñamos a evaluarse a sí mismo usando datos históricos, dándole más importancia a las tendencias de los años más recientes. Finalmente, optimizamos los tiempos de procesamiento de la computadora y, lo más importante, aislamos el impacto de la pandemia de COVID-19 marcándola como un "evento atípico" de dos años y medio; de esta forma, el modelo comprende que esa crisis fue una anomalía temporal y evita que esos datos distorsionen las predicciones del futuro.
