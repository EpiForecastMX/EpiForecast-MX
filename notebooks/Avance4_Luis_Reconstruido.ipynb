Introducción

<h3 style="color: #004C46; font-weight: 500; font-size: 1.1em;">
Diagnóstico post-entrenamiento y reconstrucción del pipeline de datos
</h3>

#### Contexto general

Una vez concluida la etapa de entrenamiento y validación de los modelos de predicción, el proyecto avanzó hacia la generación de archivos de salida y su integración en un entorno de visualización interactiva. En esta fase, comenzaron a manifestarse una serie de inconsistencias que afectaban tanto la estabilidad del pipeline como la coherencia de los resultados mostrados en Tableau.

Aunque los primeros síntomas se observaron en la capa de visualización, la persistencia de los errores incluso en vistas simplificadas indicó que el problema no era puramente visual, sino estructural, localizado en la forma en que los datos históricos y las predicciones estaban siendo integrados aguas abajo del entrenamiento.

#### Problema observado tras el entrenamiento

Tras generar los archivos de predicción, se identificaron múltiples fallas en etapas posteriores del flujo de datos: errores recurrentes en los procesos de *join* entre histórico y forecast, desaparición de fechas futuras al aplicar filtros en Tableau, inflado artificial de métricas agregadas, y fallos en validaciones del pipeline al construir el dataset final de visualización.

Estos problemas no desaparecían al simplificar dashboards ni al modificar configuraciones visuales, lo que descartó rápidamente un problema exclusivo de Tableau y apuntó a un defecto en la estructura del dataset subyacente.

<h2 style="color: #003A70; font-family: 'Segoe UI', Helvetica, Arial, sans-serif; font-weight: 300; font-size: 1.6em;">
1 &nbsp;Post-train: diagnóstico del dataset y reconstrucción del pipeline
</h2>

### 1.1 Problema raíz: duplicados en histórico (Semana 53 vs Semana 1)

El pipeline fallaba por inconsistencias en el dataset, no por Tableau. Se detectaron duplicados reales en el histórico INEGI en fechas específicas (`2014-12-29` y `2025-12-29`) por colisión de **Semana 53** vs **Semana 1**, lo que rompe la unicidad esperada para los joins.

La corrección se implementó **en el pipeline**, volviendo el histórico determinista: preferir Semana 53 cuando existe.

```python
# scripts/build_tableau.py (fragmento): fix determinista para Semana 53 vs 1

if "Semana" in real.columns:
    special_ds = pd.to_datetime(["2014-12-29", "2025-12-29"])
    is_special = real["ds"].isin(special_ds)

    if is_special.any():
        a = real.loc[~is_special].copy()
        b = real.loc[is_special].copy()

        b["_wk_pref"] = (b["Semana"] == 53).astype(int)
        b = (
            b.sort_values(
                ["padecimiento", "entidad", "ds", "_wk_pref"],
                ascending=[True, True, True, False],
            )
            .drop_duplicates(["padecimiento", "entidad", "ds"], keep="first")
            .drop(columns=["_wk_pref"])
        )
        real = pd.concat([a, b], ignore_index=True)
```

### 1.2 Política de integridad: no inventar datos (NaN explícitos)

Se decidió no imputar ni fabricar valores: no rellenar `yhat`, no mezclar valores reales con forecast, y mantener NaN como ausencia real de información.

Esto impacta directamente el diseño de `tableau.csv`, construido con `outer join` y una validación mínima correcta: **prohibir NaN solo en llaves críticas**.

### 1.3 Contrato de llaves: definición, checks y merge final

La llave compuesta del dataset final es: `ds`, `padecimiento`, `entidad`, `meta_modo`.

El pipeline exige unicidad por esa llave en **ambas** entradas (histórico expandido y forecast) antes de hacer el merge. El merge es `outer` para no perder histórico ni forecast, y se usa `validate="m:1"` para controlar cardinalidad.

```python
# scripts/build_tableau.py (fragmento): contrato de llaves y merge

join_cols = ["ds", "padecimiento", "entidad", "meta_modo"]
logger.info("Join por: {}", join_cols)

if fcst.duplicated(join_cols).any():
    raise ValueError("Duplicados en forecast para llaves de join.")
if real_long.duplicated(join_cols).any():
    raise ValueError("Duplicados en histórico expandido para llaves de join.")

out = real_long.merge(fcst, how="outer", on=join_cols, validate="m:1")
```

### 1.4 Expansión por `meta_modo`: evitar triple conteo desde datos

Al expandir el histórico a `general/hombres/mujeres`, se evitó inflar `SUM(Incrementos_*)` en Tableau dejando poblado **solo** el incremento correspondiente al modo y poniendo NaN en los otros dos incrementos.

```python
# scripts/build_tableau.py: expansión por modo sin triple conteo

def expand_real_by_modo(real: pd.DataFrame) -> pd.DataFrame:
    """
    Expande a 3 modos; solo el incremento correspondiente queda poblado,
    los demás se dejan como NaN para evitar triple conteo.
    """
    r = real.copy()

    g = r.assign(meta_modo="general", y_real=r["incrementos_total"])
    h = r.assign(meta_modo="hombres", y_real=r["incrementos_hombres"])
    m = r.assign(meta_modo="mujeres", y_real=r["incrementos_mujeres"])

    g = g.assign(incrementos_hombres=pd.NA, incrementos_mujeres=pd.NA)
    h = h.assign(incrementos_total=pd.NA, incrementos_mujeres=pd.NA)
    m = m.assign(incrementos_total=pd.NA, incrementos_hombres=pd.NA)

    return pd.concat([g, h, m], ignore_index=True)
```

### 1.5 Política de fill: rellenar numéricas generales (excluyendo incrementos)

En el dataset final se permite fill de columnas numéricas generales, **pero se bloquea explícitamente** el fill de incrementos para no alterar la semántica de conteos observados por modo.

```python
# scripts/build_tableau.py (fragmento): fill controlado de numéricas

# Solo numéricas generales, NO incrementos
block_fill = {"incrementos_total", "incrementos_hombres", "incrementos_mujeres"}
num_cols = [c for c in out.select_dtypes(include=["number"]).columns if c not in block_fill]
if num_cols:
    out[num_cols] = out[num_cols].fillna(0)
```

### 1.6 Validación mínima correcta: NaN prohibidos solo en llaves críticas

La validación final se redujo a lo mínimo correcto: el dataset puede tener NaN en valores (por diseño), pero **no puede** tener NaN en llaves críticas.

```python
# scripts/build_tableau.py (fragmento): validación de llaves críticas

key_cols = ["ds", "padecimiento", "entidad", "meta_modo"]
bad_keys = out[key_cols].isna().sum()
bad_keys = bad_keys[bad_keys > 0]
if not bad_keys.empty:
    detail = ", ".join(f"{c}: {n}" for c, n in bad_keys.items())
    raise ValueError(f"Dataset final con NaN en llaves críticas. Detalle -> {detail}")
```

Esto permite auditar: `archivo_modelo_original` vs `archivo_modelo_usado`, y comparar `mase_original` vs `mase_usado`, `confianza_original` vs `confianza_usado`.

```python
if pd.notna(row.get("usar_regional")) and row.get("confianza") == "insuficiente":
```

#### Paso A: Lectura de métricas desde `*_completo.csv` (`predice.py`)

En `predice.py`, las métricas se leen de archivos `*_completo.csv` dentro de `models/`, se concatenan y deduplican por `archivo_modelo`. Si no se descargaron, el notebook continua usando las métricas que ya vengan en `all_forecast.csv`.

````python
def cargar_metricas_completos(base_dir: Path) -> pd.DataFrame:
    frames = []
    for csv_path in sorted(base_dir.rglob("*_completo.csv")):
        try:
            df = pd.read_csv(csv_path)
        except Exception:
            continue
        if "archivo_modelo" not in df.columns:
            continue
        df["archivo_modelo"] = df["archivo_modelo"].astype(str)
        cols = [
            c for c in [
                "archivo_modelo", "rmse", "mae", "mape", "mase",
                "confianza", "normalizado", "poblacion", "usar_regional",
            ]
            if c in df.columns
        ]
        frames.append(df[cols].copy())
    if not frames:
        return pd.DataFrame(columns=["archivo_modelo"])
    return pd.concat(frames, ignore_index=True).drop_duplicates("archivo_modelo")


# Detecta ubicacion de *_completo.csv dentro del ZIP
base_models_guess = None
if "downloaded" in globals() and downloaded:
    candidates = [p for p in EXTRACT_DIR.rglob("models") if p.is_dir()]
    base_models_guess = candidates[0] if candidates else EXTRACT_DIR

BASE_MODELS = base_models_guess if base_models_guess else Path(".")
met = cargar_metricas_completos(BASE_MODELS)

if met.empty:
    print("No se encontraron *_completo.csv. Modo reproducible con all_forecast.csv.")
else:
    print(f"Metricas cargadas: {met.shape}")
    display(met.head(5))
````

#### Paso B: Construcción de `mapeo_hibrido` (fallback regional, `predice.py`)

En modo híbrido, `predice.py` construye un mapeo desde los `*_completo.csv` para identificar modelos con `confianza == "insuficiente"` y `usar_regional` no nulo.

````python
def construir_mapeo_hibrido(met_df: pd.DataFrame) -> dict:
    if met_df.empty or "usar_regional" not in met_df.columns:
        return {}
    mapeo = {}
    for _, row in met_df.iterrows():
        if pd.notna(row.get("usar_regional")) and row.get("confianza") == "insuficiente":
            stem = str(row["archivo_modelo"]).replace(".pkl", "")
            mapeo[stem] = {
                "pkl_regional": row.get("usar_regional"),
                "poblacion": row.get("poblacion"),
            }
    return mapeo

mapeo_hibrido = construir_mapeo_hibrido(met)
print("Modelos con fallback hibrido:", len(mapeo_hibrido))
list(mapeo_hibrido.items())[:3]
````

#### Paso C: Merges de métricas ORIGINAL vs USADO sobre `all_forecast` (`predice.py`)

Si `all_forecast.csv` ya trae columnas `rmse_original`, etc., el paso ya fue ejecutado en el repo. Aqui verificamos consistencia y, si existen los `*_completo.csv`, recalculamos los merges para validar que `m:1` se cumpla.

````python
cols_needed = [
    "archivo_modelo_original", "archivo_modelo_usado",
    "rmse_original", "mae_original", "mape_original", "mase_original", "confianza_original",
    "rmse_usado", "mae_usado", "mape_usado", "mase_usado", "confianza_usado",
]
present = [c for c in cols_needed if c in fcst.columns]
missing = [c for c in cols_needed if c not in fcst.columns]

print(f"Columnas metricas presentes: {len(present)}")
print(f"Columnas metricas faltantes: {len(missing)}")
if missing:
    print("Faltan:", missing[:10])

# Recalcular merges como sanity-check si tenemos met
if not met.empty and "archivo_modelo_original" in fcst.columns and "archivo_modelo_usado" in fcst.columns:
    met_base = met.copy()

    met_orig = met_base.rename(columns={
        "rmse": "rmse_original", "mae": "mae_original",
        "mape": "mape_original", "mase": "mase_original",
        "confianza": "confianza_original",
        "normalizado": "normalizado_original",
        "poblacion": "poblacion_original",
    })
    chk = fcst.merge(
        met_orig, how="left",
        left_on="archivo_modelo_original", right_on="archivo_modelo",
        validate="m:1",
    ).drop(columns=["archivo_modelo"])

    met_used = met_base.rename(columns={
        "rmse": "rmse_usado", "mae": "mae_usado",
        "mape": "mape_usado", "mase": "mase_usado",
        "confianza": "confianza_usado",
        "normalizado": "normalizado_usado",
        "poblacion": "poblacion_usado",
    })
    chk = chk.merge(
        met_used, how="left",
        left_on="archivo_modelo_usado", right_on="archivo_modelo",
        validate="m:1",
    ).drop(columns=["archivo_modelo"])

    print(f"Merge sanity-check OK. Filas: {chk.shape[0]} | Cols: {chk.shape[1]}")
    display(chk[[c for c in chk.columns if c.endswith("_original") or c.endswith("_usado")][:12]].head(3))
else:
    print("No se recalculan merges (met vacio o faltan llaves en forecast).")
````

#### Paso D: Preparación de inputs y fix Semana 53 (`build_tableau.py`)

````python
# Preparacion equivalente a prepare_inputs() en scripts/build_tableau.py
real_p = real.copy()
fcst_p = fcst.copy()

# Parse fechas + limpia invalidas
real_p["Fecha"] = pd.to_datetime(real_p["Fecha"], errors="coerce")
fcst_p["ds"] = pd.to_datetime(fcst_p["ds"], errors="coerce")

n_bad_real = int(real_p["Fecha"].isna().sum())
n_bad_fcst = int(fcst_p["ds"].isna().sum())

if n_bad_real:
    print(f"Fechas invalidas en historico (Fecha): {n_bad_real}. Se eliminan.")
if n_bad_fcst:
    print(f"Fechas invalidas en forecast (ds): {n_bad_fcst}. Se eliminan.")

# Rename historico al contrato de build_tableau.py
real_p = (
    real_p
    .dropna(subset=["Fecha"])
    .rename(columns={
        "Fecha": "ds",
        "Padecimiento": "padecimiento",
        "Entidad": "entidad",
    })
)

# Contrato forecast: ds + meta_*
fcst_p = fcst_p.dropna(subset=["ds"]).copy()
fcst_p["padecimiento"] = fcst_p["meta_padecimiento"]
fcst_p["entidad"] = fcst_p["meta_entidad"]

# Fix critico: 2014-12-29 y 2025-12-29 (Semana 53 vs Semana 1)
if "Semana" in real_p.columns:
    special_ds = pd.to_datetime(["2014-12-29", "2025-12-29"])
    is_special = real_p["ds"].isin(special_ds)

    if is_special.any():
        a = real_p.loc[~is_special].copy()
        b = real_p.loc[is_special].copy()

        b["_wk_pref"] = (b["Semana"] == 53).astype(int)
        b = (
            b.sort_values(
                ["padecimiento", "entidad", "ds", "_wk_pref"],
                ascending=[True, True, True, False],
            )
            .drop_duplicates(["padecimiento", "entidad", "ds"], keep="first")
            .drop(columns=["_wk_pref"])
        )
        real_p = pd.concat([a, b], ignore_index=True)

print("Despues de prepare_inputs:")
print(f"  real : {real_p.shape} | ds nulls: {int(real_p['ds'].isna().sum())}")
print(f"  fcst : {fcst_p.shape} | ds nulls: {int(fcst_p['ds'].isna().sum())}")

real_p.head(3)
````

#### Paso E: Expansión del histórico por `meta_modo` (sin triple conteo)

Se crean 3 copias del histórico (`general`, `hombres`, `mujeres`). Solo el incremento correspondiente al modo queda poblado; los otros se dejan como NaN para evitar inflar `SUM(Incrementos_*)` en Tableau.

````python
# Expansion por modo (equivalente a expand_real_by_modo())
r = real_p.copy()

required = ["incrementos_total", "incrementos_hombres", "incrementos_mujeres"]
missing = [c for c in required if c not in r.columns]
if missing:
    raise KeyError(f"Historico no tiene columnas requeridas para expansion: {missing}")

g = r.assign(meta_modo="general", y_real=r["incrementos_total"])
h = r.assign(meta_modo="hombres", y_real=r["incrementos_hombres"])
m = r.assign(meta_modo="mujeres", y_real=r["incrementos_mujeres"])

g = g.assign(incrementos_hombres=pd.NA, incrementos_mujeres=pd.NA)
h = h.assign(incrementos_total=pd.NA, incrementos_mujeres=pd.NA)
m = m.assign(incrementos_total=pd.NA, incrementos_hombres=pd.NA)

real_long = pd.concat([g, h, m], ignore_index=True)

join_cols = ["ds", "padecimiento", "entidad", "meta_modo"]
print(f"real_long: {real_long.shape}")
print(f"Duplicados en real_long por join_cols: {bool(real_long.duplicated(join_cols).any())}")
real_long[join_cols + ["y_real"]].head(5)
````

#### Paso F: Validación de llaves y `outer merge` histórico + forecast

````python
# Validacion de duplicados por join_cols
dup_fcst = bool(fcst_p.duplicated(join_cols).any())
dup_real = bool(real_long.duplicated(join_cols).any())

print("Duplicados por join_cols:")
print(f"  forecast:  {dup_fcst}")
print(f"  real_long: {dup_real}")

if dup_fcst:
    d = fcst_p[fcst_p.duplicated(join_cols, keep=False)].sort_values(join_cols)
    display(d.head(10))
    raise ValueError("Duplicados en forecast para llaves de join.")
if dup_real:
    d = real_long[real_long.duplicated(join_cols, keep=False)].sort_values(join_cols)
    display(d.head(10))
    raise ValueError("Duplicados en historico expandido para llaves de join.")

out = real_long.merge(fcst_p, how="outer", on=join_cols)
print(f"out (post-merge): {out.shape}")
out[join_cols + ["y_real", "yhat"]].head(8)
````

#### Paso G: Cálculo de `rmse_modelo` (solo donde hay `y_real` y `yhat`)

````python
grp = ["padecimiento", "entidad", "meta_modo"]

tmp = out[grp + ["y_real", "yhat"]].copy()
tmp = tmp[tmp["y_real"].notna() & tmp["yhat"].notna()]

tmp["y_real"] = pd.to_numeric(tmp["y_real"], errors="coerce")
tmp["yhat"] = pd.to_numeric(tmp["yhat"], errors="coerce")
tmp = tmp.dropna(subset=["y_real", "yhat"])

rmse_df = (
    tmp.assign(se=(tmp["y_real"] - tmp["yhat"]) ** 2)
       .groupby(grp, as_index=False)["se"]
       .mean()
       .assign(rmse_modelo=lambda d: d["se"] ** 0.5)
       .drop(columns=["se"])
)

out = out.merge(rmse_df, how="left", on=grp)
print(f"rmse_df: {rmse_df.shape}")
rmse_df.head(5)
````

#### Paso H: Renombres, fill controlado y validación final

````python
# Renombres minimos (si existen)
rename_map = {
    "Total": "Poblacion Total",
    "Hombres": "Poblacion Hombres",
    "Mujeres": "Poblacion Mujeres",
    "region_salud_mental": "Region Socio-Urbana",
}
out = out.rename(columns={k: v for k, v in rename_map.items() if k in out.columns})

# Fill solo numericas generales, NO incrementos
block_fill = {"incrementos_total", "incrementos_hombres", "incrementos_mujeres"}
num_cols = [c for c in out.select_dtypes(include=["number"]).columns if c not in block_fill]
if num_cols:
    out[num_cols] = out[num_cols].fillna(0)

# Validacion llaves criticas sin NaN
key_cols = ["ds", "padecimiento", "entidad", "meta_modo"]
bad_keys = out[key_cols].isna().sum()
bad_keys = bad_keys[bad_keys > 0]
print("NaN en llaves criticas:", bad_keys.to_dict())

if not bad_keys.empty:
    detail = ", ".join(f"{c}: {n}" for c, n in bad_keys.items())
    raise ValueError(f"Dataset final con NaN en llaves criticas. Detalle -> {detail}")

out = out.sort_values(["padecimiento", "entidad", "ds", "meta_modo"]).reset_index(drop=True)
out.shape
````

3 &nbsp;Tableau: migración a dataset integrado, ajustes y versionado del workbook
</h2>

### 3.1 Problema observado en Tableau (síntomas)

En las vistas de series de tiempo, las fechas futuras (p. ej., 2027) aparecían cuando se veía "All", pero desaparecían al filtrar por Entidad o Padecimiento, o al usar ciertas worksheets. Esto disparó la revisión del eje temporal, del modelo de datos (Relationships vs Join físico), del contenido real de los CSV y del efecto de filtros sobre campos no unificados.

### 3.2 Causa técnica

La causa no era el eje, sino la interacción entre un full outer join (o relaciones) y filtros aplicados sobre campos provenientes del histórico. Como el forecast futuro tiene nulos "por diseño" en campos del histórico (p. ej., cuando no existe `y_real`), ciertos filtros pueden excluir filas futuras aunque el join sea outer.

### 3.3 Decisión: migrar a un solo dataset integrado (`tableau.csv`)

**Contexto.** Inicialmente, la arquitectura de visualización se apoyaba en dos fuentes de datos: una para históricos y otra para pronósticos. Aunque funcional en etapas tempranas, este enfoque introducía complejidad innecesaria en Tableau, especialmente en la gestión de filtros, joins y consistencia de cálculos. Se decidió migrar a un **unico dataset integrado**, generado directamente por el pipeline (`scripts/build_tableau.py`).

**Nuevo enfoque.** Históricos y pronósticos conviven en un solo dominio de datos con una llave compuesta homogénea (`ds`, `padecimiento`, `entidad`, `meta_modo`). Todos los registros responden a los mismos filtros y parámetros en Tableau.

**Ajustes del lado de Tableau:**

La migración requirió reconstruir sistemáticamente múltiples componentes del workbook:

- **Reasignación de fuente de datos.** Todas las hojas, dashboards y cálculos se apuntaron al nuevo `tableau.csv`, validando que tipos de datos (fechas, categorías, medidas) fueran consistentes.
- **Revisión de cálculos.** Se actualizaron campos calculados que dependían de la separación entre históricos y pronósticos, se ajustaron condiciones lógicas basadas en `meta_modo`, y se revisaron table calculations con expresiones `WINDOW_*`.
- **Agregaciones y nivel de detalle.** Se recalibraron agregaciones (`SUM`, `AVG`, `FIXED`) para evitar duplicaciones, y se ajustaron expresiones LOD para la nueva estructura.
- **Parámetros y filtros.** Todos los parámetros se revisaron para compatibilidad con el nuevo esquema, y los filtros globales se remapearon para operar sobre una sola fuente.
- **Interacciones y dashboards.** Las acciones de filtro y resaltado se verificaron para eliminar dependencias residuales a la arquitectura anterior.

**Validación.** Se compararon visualizaciones clave contra versiones previas, se revisaron valores agregados, tendencias y cortes temporales, y se verificó la ausencia de valores nulos inesperados o rupturas en series de tiempo.

**Justificación técnica.** Aunque conceptualmente simple, la migración implicó reconstruir la lógica interna del workbook para alinearla con una nueva fuente de verdad. El resultado es una arquitectura mas limpia, mantenible y escalable, con menor fricción para análisis futuros.

### 3.4 Ajustes UX/UI

**Referencias temporales y contexto histórico.**
Se incorporaron elementos visuales que ayudan a contextualizar los datos en el tiempo: una línea de referencia asociada al periodo COVID, el ajuste del eje temporal a fecha continua (condición necesaria para bandas y líneas de referencia en Tableau), y la revisión de títulos y subtítulos para que las referencias temporales sean claras sin sobrecargar la visualización.

### 4.1 Objetivo técnico del experimento

El objetivo es **validar un principio basico de automatización**: publicar datos actualizados desde el repositorio, de forma programada, hacia una fuente consumible por Tableau Public, sin depender de abrir Tableau Desktop ni de intervención manual.

La idea central es comprobar que el flujo `pipeline reproducible > publicación automatica > visualización` funciona de extremo a extremo bajo las restricciones de Tableau Public. Si la prueba resulta exitosa, este mismo principio se aplicará a EpiForecast-MX.

### 4.2 Arquitectura del flujo end-to-end

```

  └──────────────┘                    └────────┬─────────┘

                                             │
```

Este flujo permite desacoplar completamente la actualización de datos del entorno de desarrollo visual.

**Acceso al Google Sheet.** El Sheet se comparte explícitamente con el correo de la service account con permisos de editor.

**Costos:** esta prueba no incurre en costos relevantes en Google Cloud; se mantiene dentro del free tier.

### 4.5 Resultados y limitación estructural

**Resultados de automatización.** La escritura a Google Sheets desde GitHub Actions funciona correctamente en cada ejecución. La persistencia y trazabilidad de los datos son confiables.

**Limitación estructural de Tableau Public.** La principal incertidumbre no está en la generación ni publicación de los datos, sino en el comportamiento de actualización: Tableau Public no ofrece control sobre triggers ni schedules de refresh, y la propagación de cambios no es inmediata ni determinista.

**Script de Python (`update_sheet.py`):**

```python
import os, json
from datetime import datetime, timezone
from dateutil import tz
import gspread
from gspread.exceptions import APIError
from google.oauth2.service_account import Credentials

def main():
    sa_json       = os.environ.get("GSHEET_SERVICE_ACCOUNT_JSON", "")
    spreadsheet_id = os.environ.get("SPREADSHEET_ID", "")
    sheet_name     = os.environ.get("SHEET_NAME", "Sheet1")

    print("Spreadsheet ID present:", bool(spreadsheet_id), "len:", len(spreadsheet_id))
    print("Sheet name:", sheet_name)
    print("SA json present:", bool(sa_json), "len:", len(sa_json))

    info = json.loads(sa_json)
    print("Service account:", info.get("client_email"))

    scopes = [
        "https://www.googleapis.com/auth/spreadsheets",
        "https://www.googleapis.com/auth/drive",
    ]
    creds = Credentials.from_service_account_info(info, scopes=scopes)
    gc = gspread.authorize(creds)

    try:
        sh = gc.open_by_key(spreadsheet_id)
        ws = sh.worksheet(sheet_name)

        now_utc     = datetime.now(timezone.utc)
        now_pacific = now_utc.astimezone(tz.gettz("America/Los_Angeles"))

        if ws.acell("A1").value in (None, ""):
            ws.update("A1:D1", [["idx", "last_update_utc", "last_update_pacific", "run_source"]])

        idx = max(len(ws.get_all_values()) - 1, 0) + 1

        ws.append_row(
            [idx, now_utc.isoformat(), now_pacific.strftime("%Y-%m-%d %H:%M:%S %Z"), "github-actions"],
            value_input_option="RAW",
        )
        print("OK updated:", now_utc.isoformat())

    except APIError as e:
        resp = getattr(e, "response", None)
        if resp is not None:
            print("APIError status:", getattr(resp, "status_code", "unknown"))
            txt = getattr(resp, "text", "")
            print("APIError body head:", txt[:500])
        print("Full error:", str(e))
        raise

if __name__ == "__main__":
    main()
```

5 &nbsp;Conclusiones

Este trabajo documenta y valida de manera exhaustiva el pipeline completo de EpiForecast-MX, desde el diagnóstico post-entrenamiento hasta la generación de un dataset integrado listo para visualización y automatización. El enfoque seguido prioriza trazabilidad, reproducibilidad y consistencia semántica, evitando explícitamente la invención de datos o correcciones visuales que oculten problemas estructurales.

En la fase post-train se identificó que los principales errores no provenían del modelado, sino de inconsistencias en los datos históricos y en la integración temporal, particularmente asociadas a la coexistencia de la Semana 53 y la Semana 1. La corrección determinista implementada garantiza coherencia temporal sin alterar valores observados, reforzando la integridad del histórico. Asimismo, la política explícita de NaN y la definición estricta del contrato de llaves establecen un marco sólido para detectar errores reales en lugar de encubrirlos mediante imputaciones arbitrarias.

La simulación paso a paso del pipeline permitió replicar fielmente la lógica operativa de los scripts productivos (`predice.py` y `build_tableau.py`) en un entorno reproducible, demostrando que el proceso de expansión por `meta_modo`, los merges híbridos de métricas y las validaciones finales conducen de forma consistente a la construcción de `tableau.csv`. Este ejercicio confirma que el dataset final no es un artefacto ad hoc para visualización, sino el resultado directo de reglas explícitas y verificables.

En el plano de visualización, la migración hacia un unico dataset integrado resolvió de raíz los problemas observados en Tableau, particularmente la pérdida del forecast futuro y la incoherencia entre vistas. El análisis deja claro que estos problemas no eran de UX ni de configuración superficial, sino consecuencia directa de un diseño de fuentes de datos fragmentado.

Respecto a Tableau Public, se documentó con evidencia oficial que sus limitaciones estructurales impiden la actualización automatica de la mayoría de las fuentes de datos. La unica excepción funcional es el uso de Google Sheets como backend, lo cual justifica el experimento de automatización desarrollado. Este hallazgo no se presenta como una solución general, sino como un workaround consciente de sus restricciones, util para prototipos, difusión publica y validación visual continua.

El experimento de automatización demuestra que es posible construir un flujo reproducible end-to-end mediante GitHub Actions y Google Sheets, manteniendo separación clara entre generación de datos, publicación y consumo visual. Aunque Tableau Public no está diseñado para entornos productivos, el principio arquitectónico validado aquí es directamente transferible a implementaciones mas robustas (Tableau Server, Tableau Cloud u otros stacks de BI).

En conjunto, el notebook no solo documenta un pipeline funcional, sino que actúa como evidencia metodológica de buenas prácticas en integración de datos, validación post-modelo y diseño de visualización basada en semántica y no en atajos gráficos. El trabajo deja al proyecto en un estado listo para evaluación académica, extensión productiva o migración a infraestructura de mayor escala.
